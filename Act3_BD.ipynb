```python
%%sql

PRAGMA foreign_keys = ON;

-- Creación de Tablas
CREATE TABLE marcas (
    id_marca INTEGER PRIMARY KEY AUTOINCREMENT,
    nombre TEXT NOT NULL
);

CREATE TABLE modelos (
    id_modelo INTEGER PRIMARY KEY AUTOINCREMENT,
    id_marca INTEGER NOT NULL,
    nombre_modelo TEXT NOT NULL,
    anio INTEGER NOT NULL,
    tipo_motor TEXT NOT NULL,
    potencia TEXT NOT NULL,
    num_cilindros INTEGER NOT NULL,
    precio_base REAL NOT NULL,
    descuento REAL NOT NULL DEFAULT 0.00,
    FOREIGN KEY (id_marca) REFERENCES marcas(id_marca) ON DELETE CASCADE
);

CREATE TABLE equipamientos (
    id_equipamiento INTEGER PRIMARY KEY AUTOINCREMENT,
    nombre_caracteristica TEXT NOT NULL
);

CREATE TABLE equipamiento_modelo (
    id_modelo INTEGER NOT NULL,
    id_equipamiento INTEGER NOT NULL,
    es_de_serie INTEGER NOT NULL,
    precio_extra REAL NOT NULL DEFAULT 0.00,
    PRIMARY KEY (id_modelo, id_equipamiento),
    FOREIGN KEY (id_modelo) REFERENCES modelos(id_modelo) ON DELETE CASCADE,
    FOREIGN KEY (id_equipamiento) REFERENCES equipamientos(id_equipamiento) ON DELETE CASCADE
);

CREATE TABLE ubicaciones (
    id_ubicacion INTEGER PRIMARY KEY AUTOINCREMENT,
    nombre_lugar TEXT NOT NULL,
    tipo_lugar TEXT CHECK(tipo_lugar IN ('Local de Ventas', 'Sucursal', 'Almacén de Resguardo', 'Servicio Oficial')) NOT NULL,
    domicilio TEXT NOT NULL,
    telefono TEXT NOT NULL
);

CREATE TABLE autos_stock (
    numero_serie TEXT PRIMARY KEY,
    id_modelo INTEGER NOT NULL,
    id_ubicacion INTEGER NOT NULL,
    matricula TEXT NOT NULL,
    es_stock INTEGER NOT NULL,
    FOREIGN KEY (id_modelo) REFERENCES modelos(id_modelo) ON DELETE CASCADE,
    FOREIGN KEY (id_ubicacion) REFERENCES ubicaciones(id_ubicacion) ON DELETE CASCADE
);

CREATE TABLE vendedores (
    id_vendedor INTEGER PRIMARY KEY AUTOINCREMENT,
    nombre TEXT NOT NULL,
    domicilio TEXT NOT NULL,
    telefono TEXT NOT NULL,
    whatsapp TEXT NOT NULL
);

CREATE TABLE ventas_acumuladas (
    id_venta INTEGER PRIMARY KEY AUTOINCREMENT,
    numero_serie TEXT NOT NULL,
    id_vendedor INTEGER NULL,
    id_ubicacion INTEGER NOT NULL,
    fecha_venta TEXT NOT NULL,
    fecha_entrega TEXT NOT NULL,
    precio_venta REAL NOT NULL,
    modo_pago TEXT CHECK(modo_pago IN ('Contado', 'Crédito')) NOT NULL,
    de_stock INTEGER NOT NULL,
    FOREIGN KEY (numero_serie) REFERENCES autos_stock(numero_serie),
    FOREIGN KEY (id_vendedor) REFERENCES vendedores(id_vendedor),
    FOREIGN KEY (id_ubicacion) REFERENCES ubicaciones(id_ubicacion)
);


-- Marcas
INSERT INTO marcas (nombre) VALUES
('Audi'), ('Porsche'), ('BMW'), ('Mercedes-Benz'), ('Lexus'), ('Maserati'), ('Bentley'), ('Aston Martin'), ('Lamborghini'), ('Ferrari'),
('Rolls-Royce'), ('Jaguar'), ('Land Rover'), ('Alfa Romeo'), ('Cadillac'), ('Genesis'), ('McLaren'), ('Bugatti'), ('Lotus'), ('Tesla');

-- Modelos
INSERT INTO modelos (id_marca, nombre_modelo, anio, tipo_motor, potencia, num_cilindros, precio_base, descuento) VALUES
(1, 'R8 V10', 2023, 'Gasolina V10', '610 HP', 10, 180000.00, 5.00),
(1, 'A4 Base', 2018, 'Turbo I4', '200 HP', 4, 35000.00, 0.00),
(2, '911 Carrera', 2024, 'Twin-Turbo Flat-6', '379 HP', 6, 120000.00, 0.00),
(2, 'Boxster', 2019, 'Turbo Flat-4', '300 HP', 4, 55000.00, 2.00),
(3, 'M5 Competition', 2022, 'Twin-Turbo V8', '617 HP', 8, 110000.00, 3.00),
(4, 'AMG GT', 2023, 'Biturbo V8', '523 HP', 8, 140000.00, 0.00),
(5, 'LC 500', 2021, 'V8 Atmosférico', '471 HP', 8, 98000.00, 4.00),
(6, 'MC20', 2023, 'Nettuno V6', '621 HP', 6, 210000.00, 0.00),
(7, 'Continental GT', 2022, 'Twin-Turbo W12', '650 HP', 12, 240000.00, 0.00),
(8, 'Vantage', 2021, 'Twin-Turbo V8', '503 HP', 8, 145000.00, 5.00),
(9, 'Huracán EVO', 2023, 'V10 Atmosférico', '631 HP', 10, 260000.00, 0.00),
(10, 'F8 Tributo', 2022, 'Twin-Turbo V8', '710 HP', 8, 280000.00, 0.00),
(11, 'Phantom', 2023, 'Twin-Turbo V12', '563 HP', 12, 450000.00, 0.00),
(12, 'F-Type R', 2020, 'Supercharged V8', '575 HP', 8, 105000.00, 8.00),
(13, 'Range Rover Sport', 2023, 'Mild Hybrid I6', '395 HP', 6, 90000.00, 0.00),
(14, 'Giulia Quadrifoglio', 2021, 'Twin-Turbo V6', '505 HP', 6, 80000.00, 5.00),
(15, 'Escalade V', 2023, 'Supercharged V8', '682 HP', 8, 150000.00, 0.00),
(16, 'G90', 2022, 'Twin-Turbo V6', '375 HP', 6, 88000.00, 3.00),
(17, '720S', 2022, 'Twin-Turbo V8', '710 HP', 8, 300000.00, 0.00),
(18, 'Chiron', 2021, 'Quad-Turbo W16', '1479 HP', 16, 3000000.00, 0.00);

-- Equipamientos
INSERT INTO equipamientos (nombre_caracteristica) VALUES
('Airbag Conductor'), ('Cierre Automático'), ('Airbag Acompañante'), ('Aire Acondicionado'), ('Pintura Metalizada'),
('Quemacocos Panorámico'), ('Asientos de Piel Nappa'), ('Frenos Cerámicos'), ('Sistema de Sonido Burmester'), ('Control de Crucero Adaptativo'),
('Cámara 360 Grados'), ('Suspensión Neumática'), ('Head-Up Display'), ('Faros Matrix LED'), ('Asistente de Carril'),
('Escape Deportivo'), ('Volante en Alcántara'), ('Cargador Inalámbrico'), ('Llantas Forjadas 21 pulgadas'), ('Visión Nocturna');

-- Equipamiento_Modelo
INSERT INTO equipamiento_modelo (id_modelo, id_equipamiento, es_de_serie, precio_extra) VALUES
(1, 1, 1, 0.00), (1, 8, 0, 8500.00), (1, 16, 0, 3200.00),
(3, 1, 1, 0.00), (3, 6, 0, 2500.00), (3, 7, 1, 0.00),
(5, 2, 1, 0.00), (5, 9, 0, 4000.00), (6, 5, 0, 1800.00),
(8, 1, 1, 0.00), (8, 8, 0, 9000.00), (9, 7, 1, 0.00),
(11, 1, 1, 0.00), (11, 19, 0, 6000.00), (12, 16, 0, 3500.00),
(13, 6, 1, 0.00), (15, 7, 1, 0.00), (17, 8, 1, 0.00),
(18, 8, 1, 0.00), (20, 20, 0, 5000.00);

-- Ubicaciones
INSERT INTO ubicaciones (nombre_lugar, tipo_lugar, domicilio, telefono) VALUES
('Concesionaria San Pedro', 'Local de Ventas', 'Av. Vasconcelos 100', '8111111101'),
('Sucursal Santa Lucía', 'Sucursal', 'Av. Revolución 450', '8111111102'),
('Almacén Central Norte', 'Almacén de Resguardo', 'Carretera a Laredo Km 12', '8111111103'),
('Servicio Oficial Premier', 'Servicio Oficial', 'Av. Constitución 890', '8111111104'),
('Concesionaria Guadalajara', 'Local de Ventas', 'Av. Patria 500', '3331111105'),
('Sucursal Zapopan', 'Sucursal', 'Av. Vallarta 1200', '3331111106'),
('Almacén Occidente', 'Almacén de Resguardo', 'Periférico Sur 300', '3331111107'),
('Servicio Oficial Elite GT', 'Servicio Oficial', 'Av. López Mateos 45', '3331111108'),
('Concesionaria Polanco', 'Local de Ventas', 'Av. Masaryk 200', '5551111109'),
('Sucursal Santa Fe', 'Sucursal', 'Av. Vasco de Quiroga 50', '5551111110'),
('Almacén Vallejo', 'Almacén de Resguardo', 'Calzada Vallejo 900', '5551111111'),
('Servicio Oficial Pedregal', 'Servicio Oficial', 'Anillo Periférico 3000', '5551111112'),
('Concesionaria Querétaro', 'Local de Ventas', 'Av. Bernardo Quintana 10', '4421111113'),
('Sucursal Juriquilla', 'Sucursal', 'Av. de las Ciencias 12', '4421111114'),
('Almacén Bajío', 'Almacén de Resguardo', 'Carretera Querétaro-SLA Km 5', '4421111115'),
('Servicio Oficial El Marqués', 'Servicio Oficial', 'Av. del Parque 4', '4421111116'),
('Concesionaria Puebla', 'Local de Ventas', 'Vía Atlixcáyotl 1500', '2221111117'),
('Sucursal Angelópolis', 'Sucursal', 'Av. Osa Mayor 30', '2221111118'),
('Almacén Puebla Sur', 'Almacén de Resguardo', 'Periférico Ecológico 80', '2221111119'),
('Servicio Oficial Las Ánimas', 'Servicio Oficial', 'Blvd. Atlixco 22', '2221111120');

-- Vendedores
INSERT INTO vendedores (nombre, domicilio, telefono, whatsapp) VALUES
('Carlos Mendoza', 'Calle Olmos 123', '8120000001', '8120000001'),
('Ana Sofía Torres', 'Av. Palmas 456', '8120000002', '8120000002'),
('Luis Fernando Gómez', 'Calle Pino 789', '8120000003', '8120000003'),
('Mariana Ríos', 'Av. Central 101', '8120000004', '8120000004'),
('Roberto Garza', 'Calle Naranjo 202', '3330000005', '3330000005'),
('Valeria Hernández', 'Av. Hidalgo 303', '3330000006', '3330000006'),
('Diego Navarro', 'Calle Roble 404', '3330000007', '3330000007'),
('Claudia Morales', 'Av. Juárez 505', '3330000008', '3330000008'),
('Alejandro Silva', 'Calle Cedro 606', '5550000009', '5550000009'),
('Sofia Castillo', 'Av. Reforma 707', '5550000010', '5550000010'),
('Gabriel Ortiz', 'Calle Fresno 808', '5550000011', '5550000011'),
('Daniela Vega', 'Av. Insurgentes 909', '5550000012', '5550000012'),
('Javier Sotomayor', 'Calle Álamos 111', '4420000013', '4420000013'),
('Lucía Méndez', 'Av. Zaragoza 222', '4420000014', '4420000014'),
('Esteban Luna', 'Calle Nogal 333', '4420000015', '4420000015'),
('Patricia Delgado', 'Av. Constituyentes 444', '4420000016', '4420000016'),
('Ricardo Paredes', 'Calle Abedul 555', '2220000017', '2220000017'),
('Andrea Benítez', 'Av. 3 Poniente 666', '2220000018', '2220000018'),
('Fernando Ibarra', 'Calle Sauce 777', '2220000019', '2220000019'),
('Camila Roldán', 'Av. 5 de Mayo 888', '2220000020', '2220000020');

-- Autos en Stock
INSERT INTO autos_stock (numero_serie, id_modelo, id_ubicacion, matricula, es_stock) VALUES
('VIN001', 1, 1, 'NLE-101', 1), ('VIN002', 3, 1, 'NLE-102', 1),
('VIN003', 5, 1, 'NLE-103', 0), ('VIN004', 6, 2, 'NLE-104', 1),
('VIN005', 7, 4, 'NLE-105', 1), ('VIN006', 8, 5, 'JAL-201', 1),
('VIN007', 9, 5, 'JAL-202', 1), ('VIN008', 10, 6, 'JAL-203', 0),
('VIN009', 11, 8, 'JAL-204', 1), ('VIN010', 12, 9, 'MEX-301', 1),
('VIN011', 13, 9, 'MEX-302', 1), ('VIN012', 14, 10, 'MEX-303', 0),
('VIN013', 15, 12, 'MEX-304', 1), ('VIN014', 16, 13, 'QRO-401', 1),
('VIN015', 17, 13, 'QRO-402', 1), ('VIN016', 18, 14, 'QRO-403', 1),
('VIN017', 19, 16, 'QRO-404', 0), ('VIN018', 20, 17, 'PUE-501', 1),
('VIN019', 1, 17, 'PUE-502', 1), ('VIN020', 3, 18, 'PUE-503', 1),
('VIN021', 5, 20, 'PUE-504', 1), ('VIN022', 6, 1, 'NLE-106', 1),
('VIN023', 8, 5, 'JAL-205', 0), ('VIN024', 9, 9, 'MEX-305', 1),
('VIN025', 11, 13, 'QRO-405', 1);

-- Ventas Acumuladas
INSERT INTO ventas_acumuladas (numero_serie, id_vendedor, id_ubicacion, fecha_venta, fecha_entrega, precio_venta, modo_pago, de_stock) VALUES
('VIN001', 1, 1, '2026-01-10', '2026-01-15', 171000.00, 'Contado', 1),
('VIN002', 2, 1, '2026-01-12', '2026-01-18', 120000.00, 'Crédito', 1),
('VIN003', 3, 1, '2026-01-15', '2026-02-01', 106700.00, 'Contado', 0),
('VIN004', 4, 2, '2026-01-18', '2026-01-22', 140000.00, 'Crédito', 1),
('VIN005', NULL, 4, '2026-01-20', '2026-01-25', 94080.00, 'Contado', 1),
('VIN006', 5, 5, '2026-01-22', '2026-01-28', 210000.00, 'Contado', 1),
('VIN007', 6, 5, '2026-01-25', '2026-01-30', 240000.00, 'Crédito', 1),
('VIN008', 7, 6, '2026-02-01', '2026-02-20', 137750.00, 'Contado', 0),
('VIN009', NULL, 8, '2026-02-03', '2026-02-08', 260000.00, 'Crédito', 1),
('VIN010', 9, 9, '2026-02-05', '2026-02-10', 280000.00, 'Contado', 1),
('VIN011', 10, 9, '2026-02-08', '2026-02-14', 450000.00, 'Crédito', 1),
('VIN012', 11, 10, '2026-02-10', '2026-03-01', 96600.00, 'Contado', 0),
('VIN013', NULL, 12, '2026-02-12', '2026-02-16', 90000.00, 'Contado', 1),
('VIN014', 13, 13, '2026-02-15', '2026-02-20', 76000.00, 'Crédito', 1),
('VIN015', 14, 13, '2026-02-18', '2026-02-22', 150000.00, 'Contado', 1),
('VIN016', 15, 14, '2026-02-20', '2026-02-25', 85360.00, 'Crédito', 1),
('VIN017', NULL, 16, '2026-02-22', '2026-03-15', 300000.00, 'Contado', 0),
('VIN018', 17, 17, '2026-02-25', '2026-03-01', 3000000.00, 'Crédito', 1),
('VIN019', 18, 17, '2026-02-26', '2026-03-02', 171000.00, 'Contado', 1),
('VIN020', 19, 18, '2026-02-27', '2026-03-03', 120000.00, 'Crédito', 1),
('VIN021', NULL, 20, '2026-02-28', '2026-03-05', 106700.00, 'Contado', 1),
('VIN022', 1, 1, '2026-03-01', '2026-03-05', 140000.00, 'Crédito', 1),
('VIN023', 5, 5, '2026-03-01', '2026-03-20', 210000.00, 'Contado', 0),
('VIN024', 9, 9, '2026-03-02', '2026-03-07', 240000.00, 'Crédito', 1),
('VIN025', 13, 13, '2026-03-02', '2026-03-06', 260000.00, 'Contado', 1);



```

```python
%%sql

SELECT
    v.id_venta,
    v.numero_serie,
    v.fecha_venta,
    v.precio_venta,
    v.modo_pago,
    u.nombre_lugar AS concesionaria
FROM ventas_acumuladas v
JOIN ubicaciones u ON v.id_ubicacion = u.id_ubicacion
WHERE u.nombre_lugar = 'Concesionaria San Pedro'
LIMIT 5;
```

```python
%%sql

SELECT
    v.id_venta,
    v.numero_serie,
    v.fecha_venta,
    v.precio_venta,
    v.modo_pago,
    u.nombre_lugar AS concesionaria
FROM ventas_acumuladas v
JOIN ubicaciones u ON v.id_ubicacion = u.id_ubicacion
WHERE v.precio_venta > 150000.00
ORDER BY v.precio_venta DESC;
```

```python
%%sql
SELECT
    id_ubicacion,
    nombre_lugar AS nombre_original,
    REPLACE(nombre_lugar, nombre_lugar, nombre_lugar || '-Nacional') AS nombre_modificado
FROM ubicaciones
WHERE tipo_lugar = 'Local de Ventas';
```

```python
%%sql
SELECT
    v.id_venta,
    v.numero_serie,
    v.fecha_venta,
    v.precio_venta,
    v.modo_pago,
    u.nombre_lugar AS concesionaria
FROM ventas_acumuladas v
JOIN ubicaciones u ON v.id_ubicacion = u.id_ubicacion
WHERE u.nombre_lugar = 'Concesionaria San Pedro-Nacional'
LIMIT 5;
```
